```python
import cv2

video_path = "skinning/interpolation_code/videos/Apple.mp4"

cap = cv2.VideoCapture(video_path)

frame_id = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    print(frame_id, frame.shape)

    # frame is a NumPy array (H, W, 3)

    frame_id += 1

cap.release()
```
